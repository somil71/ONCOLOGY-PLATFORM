```
"""Validated manifest selection and geometric ordering for research image inputs.

Raster identity must come from a reviewed manifest, never from its extension.
DICOM sequences must be selected explicitly before ordering; this code does not
choose clinically appropriate MRI protocols or establish prediction-time eligibility.
"""

import csv
from pathlib import Path
import numpy as np


def load_image_manifest(manifest_path, image_root):
    root = Path(image_root).resolve()
    groups = {}
    seen = set()
    with Path(manifest_path).open(encoding="utf-8-sig", newline="") as stream:
        reader = csv.DictReader(stream)
        if not {"patient_id", "modality", "relative_path", "identity_source"} <= set(
            reader.fieldnames or []
        ):
            raise ValueError(
                "Image manifest requires patient_id, modality, relative_path, identity_source."
            )
        for row in reader:
            patient, modality = (
                row["patient_id"].strip(),
                row["modality"].strip().lower(),
            )
            if not patient or any(token in patient for token in ("/", "\\", "..")):
                raise ValueError("Invalid manifest patient identifier.")
            if modality not in ("wsi", "rad") or not row["identity_source"].strip():
                raise ValueError(
                    "Each image needs a reviewed modality and identity source."
                )
            relative = Path(row["relative_path"])
            path = (root / relative).resolve()
            if (
                relative.is_absolute()
                or not path.is_relative_to(root)
                or not path.is_file()
            ):
                raise ValueError("Manifest image must exist inside the image root.")
            if path in seen:
                raise ValueError(
                    "An image cannot be assigned twice or to multiple patients/modalities."
                )
            if modality == "rad":
                import pydicom

                header = pydicom.dcmread(str(path), stop_before_pixels=True)
                if str(header.get("PatientID", "")).strip() != patient:
                    raise ValueError(
                        "Radiology PatientID must match the reviewed case mapping."
                    )
            seen.add(path)
            groups.setdefault((patient, modality), []).append(path)
    return groups


def order_dicom_series(paths, modality):
    """Check one selected MR/CT series and sort by physical slice position."""
    if modality not in ("MR", "CT"):
        raise ValueError("Only reviewed MR or CT volumes are supported.")
    import pydicom

    if not paths:
        return []
    records = []
    for path in paths:
        ds = pydicom.dcmread(str(path), stop_before_pixels=True)
        if ds.get("Modality") != modality or int(ds.get("NumberOfFrames", 1)) != 1:
            raise ValueError(
                "Expected single-frame matching-modality DICOM; enhanced/multiframe needs a dedicated loader."
            )
        for name in (
            "PatientID",
            "SeriesInstanceUID",
            "FrameOfReferenceUID",
            "ImageOrientationPatient",
            "ImagePositionPatient",
            "Rows",
            "Columns",
            "PixelSpacing",
        ):
            if name not in ds:
                raise ValueError(f"DICOM geometry is missing {name}.")
        for name in ("PatientID", "SeriesInstanceUID", "FrameOfReferenceUID"):
            if not str(ds.get(name, "")).strip():
                raise ValueError(f"DICOM identity is empty: {name}.")
        if int(ds.Rows) <= 0 or int(ds.Columns) <= 0:
            raise ValueError("DICOM Rows and Columns must be positive.")
        orientation = np.asarray(ds.ImageOrientationPatient, dtype=float)
        position = np.asarray(ds.ImagePositionPatient, dtype=float)
        spacing = np.asarray(ds.PixelSpacing, dtype=float)
        if (
            orientation.shape != (6,)
            or position.shape != (3,)
            or spacing.shape != (2,)
            or not np.isfinite(np.r_[orientation, position, spacing]).all()
            or (spacing <= 0).any()
        ):
            raise ValueError("Invalid DICOM geometry.")
        if (
            not np.isclose(np.linalg.norm(orientation[:3]), 1, atol=1e-4)
            or not np.isclose(np.linalg.norm(orientation[3:]), 1, atol=1e-4)
            or not np.isclose(np.dot(orientation[:3], orientation[3:]), 0, atol=1e-4)
        ):
            raise ValueError(
                "DICOM orientation must contain orthonormal direction cosines."
            )
        records.append((Path(path), ds, orientation, spacing, position))
    first = records[0]
    reference_normal = np.cross(first[2][:3], first[2][3:])
    reference_normal /= np.linalg.norm(reference_normal)
    for record in records[1:]:
        if (
            str(record[1].PatientID) != str(first[1].PatientID)
            or str(record[1].SeriesInstanceUID) != str(first[1].SeriesInstanceUID)
            or str(record[1].FrameOfReferenceUID) != str(first[1].FrameOfReferenceUID)
        ):
            raise ValueError("Select one DICOM series and frame of reference per bag.")
        if (
            (int(record[1].Rows), int(record[1].Columns))
            != (int(first[1].Rows), int(first[1].Columns))
            or not np.allclose(record[2], first[2], atol=1e-4)
            or not np.allclose(record[3], first[3], atol=1e-4)
        ):
            raise ValueError("Mixed DICOM orientation, dimensions or pixel spacing.")
        delta = record[4] - first[4]
        in_plane_shift = delta - float(delta @ reference_normal) * reference_normal
        if np.linalg.norm(in_plane_shift) > 1e-3:
            raise ValueError(
                "DICOM slice origins shift within the image plane; resample explicitly before volume feature extraction."
            )
    records.sort(key=lambda record: float(record[4] @ reference_normal))
    gaps = np.diff([float(record[4] @ reference_normal) for record in records])
    if (gaps <= 1e-4).any():
        raise ValueError(
            "Duplicate or overlapping slice positions; select a single acquisition."
        )
    if len(gaps) > 1 and not np.allclose(gaps, np.median(gaps), rtol=0.05, atol=0.1):
        raise ValueError(
            "Nonuniform slice spacing; resample explicitly before feature extraction."
        )
    return [record[0] for record in records]


def select_evenly_spaced(paths, limit):
    if limit < 1:
        raise ValueError("Image item limit must be positive.")
    if len(paths) <= limit:
        return list(paths)
    return [paths[index] for index in np.linspace(0, len(paths) - 1, limit, dtype=int)]


def order_mr_series(paths):
    """Validate selected MR data without inferring sequence identity."""
    return order_dicom_series(paths, "MR")


def order_ct_series(paths):
    """Validate selected CT data; intensity conversion is a separate contract."""
    return order_dicom_series(paths, "CT")


"""Audit the Phase 2 source tables and the image loader's actual file selection."""

from collections import Counter
import json
from pathlib import Path

import numpy as np
import pandas as pd


def audit_outcomes(demographics, output_dir):
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)
    columns = {
        "status": "demographic_vital_status",
        "death": "demographic_days_to_death",
        "followup": "diagnoses_days_to_last_follow_up",
    }
    missing = [column for column in columns.values() if column not in demographics]
    summary = {"source_rows": len(demographics), "required_columns_missing": missing}
    status_column = columns["status"]
    if status_column in demographics:
        summary["raw_status_counts"] = {
            str(key): int(value)
            for key, value in demographics[status_column].value_counts(dropna=False).items()
        }
    if not missing:
        rows = demographics[["Patient_ID", *columns.values()]].copy()
        status = rows[status_column].astype("string").str.strip().str.casefold()
        rows["event"] = status.map({"alive": 0, "living": 0, "dead": 1, "deceased": 1})
        death = pd.to_numeric(rows[columns["death"]], errors="coerce")
        followup = pd.to_numeric(rows[columns["followup"]], errors="coerce")
        rows["time"] = death.where(rows["event"].eq(1), followup)
        rows["usable"] = rows["event"].notna() & np.isfinite(rows["time"]) & rows["time"].gt(0)
        rows["unmapped_status"] = rows["event"].isna()
        rows["invalid_duration"] = ~np.isfinite(rows["time"]) | rows["time"].le(0)
        rows.to_csv(output_dir / "outcome_rows.csv", index=False)
        summary.update({
            "usable_rows": int(rows["usable"].sum()),
            "usable_event_rows": int((rows["usable"] & rows["event"].eq(1)).sum()),
            "unmapped_status_rows": int(rows["unmapped_status"].sum()),
            "invalid_duration_rows": int(rows["invalid_duration"].sum()),
            "duplicate_patient_rows": int(rows["Patient_ID"].duplicated().sum()),
        })
    (output_dir / "outcome_summary.json").write_text(json.dumps(summary, indent=2), encoding="utf-8")
    return summary


def audit_modality_files(cohort, image_root, rna_ids, clinical_ids, max_items, output_dir, image_manifest=None):
    """Replay the current extension-based routing and decoding without neural inference.

    The manifest describes loader behavior. It does not validate imaging modality identity.
    """
    from PIL import Image
    import pydicom

    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)
    rna_ids, clinical_ids = set(rna_ids), set(clinical_ids)
    cases, files = [], []
    for record in cohort.to_dict("records"):
        patient = str(record["Patient_ID"])
        # Patient identifiers are directory names, never paths supplied to the file system.
        if Path(patient).name != patient or patient in (".", "..") or "/" in patient or "\\" in patient:
            raise ValueError(f"Invalid patient directory identifier: {patient!r}")
        candidates = sorted(path for path in (Path(image_root) / patient).rglob("*") if path.is_file())
        if image_manifest is None:
            # Historical loader audit retained to explain archived runs only.
            wsi = [path for path in candidates if path.suffix in (".jpg", ".png")]
            rad = [path for path in candidates if path.suffix not in (".jpg", ".png")]
        else:
            wsi = sorted(image_manifest.get((patient, 'wsi'), []))
            rad = order_mr_series(image_manifest.get((patient, 'rad'), []))
            wsi, rad = select_evenly_spaced(wsi, max_items), select_evenly_spaced(rad, max_items)
        counts = Counter()
        for modality, paths in (("wsi", wsi), ("rad", rad)):
            for path in paths[:max_items]:
                error = ""
                try:
                    if modality == "wsi":
                        with Image.open(path) as im:
                            im.convert("RGB").resize((224, 224)).load()
                    else:
                        array = pydicom.dcmread(str(path), force=True).pixel_array.astype(float)
                        array = (array - np.min(array)) / (np.max(array) - np.min(array) + 1e-8)
                        array = (array * 255).astype(np.uint8)
                        Image.fromarray(array if array.ndim == 2 else array[0]).convert("RGB").resize((224, 224)).load()
                    counts[modality] += 1
                except Exception as exc:
                    error = f"{type(exc).__name__}: {exc}"
                files.append({
                    "Patient_ID": patient, "loader_branch": modality,
                    "relative_path": path.relative_to(image_root).as_posix(),
                    "decoded": not error, "error": error,
                })
        cases.append({
            **record, "wsi_candidates": len(wsi), "rad_candidates": len(rad),
            "wsi_decoded": counts["wsi"], "rad_decoded": counts["rad"],
            "rna_available": patient in rna_ids, "cli_available": patient in clinical_ids,
            "raster_in_radiology_path": sum(
                any(word in path.as_posix().lower() for word in ("mri", "radiology", "dicom")) for path in wsi
            ),
        })
    frame = pd.DataFrame(cases)
    frame.to_csv(output_dir / "modality_manifest.csv", index=False)
    pd.DataFrame(files, columns=["Patient_ID", "loader_branch", "relative_path", "decoded", "error"]).to_csv(
        output_dir / "image_decode_audit.csv", index=False
    )
    partitions = {}
    for split, rows in frame.groupby("split"):
        partitions[split] = {
            "cases": len(rows), "events": int(rows["event"].sum()),
            "wsi_usable_cases": int(rows["wsi_decoded"].gt(0).sum()),
            "rad_usable_cases": int(rows["rad_decoded"].gt(0).sum()),
            "rna_available_cases": int(rows["rna_available"].sum()),
            "cli_available_cases": int(rows["cli_available"].sum()),
            "all_four_available_cases": int((rows["wsi_decoded"].gt(0) & rows["rad_decoded"].gt(0)
                                              & rows["rna_available"] & rows["cli_available"]).sum()),
        }
    summary = {
        "image_root": str(image_root), "max_items_tested_per_branch": max_items,
        "routing": "Reviewed modality manifest; geometric MR series ordering." if image_manifest is not None else "Historical loader assigns lowercase jpg/png to WSI and other files to RAD.",
        "interpretation": "Decode success does not establish true modality identity or feature quality.",
        "partitions": partitions,
    }
    (output_dir / "modality_summary.json").write_text(json.dumps(summary, indent=2), encoding="utf-8")
    return summary


class PanCancerDataset(Dataset):
    def __init__(self, df, rna_df, cli_df):
        self.df = df
        self.rna_df = rna_df
        self.cli_df = cli_df
        self.feature_cache = {}
        if not CFG.get('image_manifest_csv'):
            raise ValueError('Set image_manifest_csv to a reviewed manifest; extension-based image identity is disabled.')
        self.img_folder = CFG['dir_multimodal']
        self.image_manifest = load_image_manifest(CFG['image_manifest_csv'], self.img_folder)

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        if not CFG['cache_frozen_features']:
            return self._encode_case(idx)
        if idx not in self.feature_cache:
            self.feature_cache[idx] = self._encode_case(idx)
        # Default collation stacks these CPU tensors into separate batch buffers.
        return self.feature_cache[idx]

    def _encode_case(self, idx):
        patient_id = self.df.iloc[idx]['Patient_ID']
        
        # Initialize Tensors (1 WSI, 2 RAD, 3 RNA, 4 CLI)
        wsi_tensor = torch.zeros((CFG['max_patches'], 768))
        rad_tensor = torch.zeros((CFG['max_patches'], 512))
        wsi_valid = torch.zeros(CFG['max_patches'], dtype=torch.bool)
        rad_valid = torch.zeros(CFG['max_patches'], dtype=torch.bool)
        mask = torch.zeros(4) # [WSI, RAD, RNA, CLI]
        
        # 1. Image Modalities (WSI & Radiology)
        wsi_files = select_evenly_spaced(sorted(self.image_manifest.get((str(patient_id), 'wsi'), [])), CFG['max_patches'])
        rad_files = select_evenly_spaced(order_mr_series(self.image_manifest.get((str(patient_id), 'rad'), [])), CFG['max_patches'])

        # Process WSI
        if len(wsi_files) > 0:
            loaded = []
            for p in wsi_files[:CFG['max_patches']]:
                try:
                    with Image.open(p) as image:
                        loaded.append(img_transform(image.convert('RGB')))
                except: pass
            if loaded:
                mask[0] = 1.0
                wsi_valid[:len(loaded)] = True
                with torch.no_grad():
                    wsi_tensor[:len(loaded)] = wsi_model(torch.stack(loaded).to(DEVICE)).cpu()
                    
        # Process Radiology (DICOM)
        if len(rad_files) > 0:
            loaded = []
            for p in rad_files[:CFG['max_patches']]:
                try:
                    ds = pydicom.dcmread(p, stop_before_pixels=False, force=True)
                    arr = ds.pixel_array.astype(float)
                    arr = (arr - np.min(arr)) / (np.max(arr) - np.min(arr) + 1e-8)
                    arr = (arr * 255).astype(np.uint8)
                    img = Image.fromarray(arr if len(arr.shape)==2 else arr[0]).convert('RGB')
                    loaded.append(img_transform(img))
                except: pass
            if loaded:
                mask[1] = 1.0
                rad_valid[:len(loaded)] = True
                with torch.no_grad():
                    rad_tensor[:len(loaded)] = rad_model(torch.stack(loaded).to(DEVICE)).cpu()

        # 2 & 3. RNA & Clinical
        rna = torch.tensor(self.rna_df.loc[patient_id].values, dtype=torch.float32) if patient_id in self.rna_df.index else torch.zeros(self.rna_df.shape[1])
        cli = torch.tensor(self.cli_df.loc[patient_id].values, dtype=torch.float32) if patient_id in self.cli_df.index else torch.zeros(self.cli_df.shape[1])
        if patient_id in self.rna_df.index: mask[2] = 1.0
        if patient_id in self.cli_df.index: mask[3] = 1.0
        
        # Target
        time = self.df.iloc[idx]['time']
        event = self.df.iloc[idx]['event']
        
        return {'patient_id': str(patient_id), 'wsi': wsi_tensor, 'rad': rad_tensor, 'rna': rna, 'cli': cli, 'mask': mask, 'wsi_valid': wsi_valid, 'rad_valid': rad_valid, 'time': time, 'event': event}
```

```
import torch
import torch.nn as nn
import torch.nn.functional as F


def _item_mask(x, valid_items=None):
    """Validate bag membership; infer legacy zero padding when unspecified."""
    if x.ndim != 3 or x.shape[1] == 0:
        raise ValueError("Image features require nonempty [batch, items, features].")
    valid = x.ne(0).any(dim=-1) if valid_items is None else valid_items
    if valid.shape != x.shape[:2] or not torch.all((valid == 0) | (valid == 1)):
        raise ValueError("Item mask must be binary with shape [batch, items].")
    valid = valid.bool()
    if not torch.isfinite(x.masked_select(valid.unsqueeze(-1).expand_as(x))).all():
        raise ValueError("Available image features must be finite.")
    return valid


def _safe_attention_mask(valid):
    safe = valid.clone()
    safe[~safe.any(dim=1), 0] = True
    return safe


# =====================================================================
# 🔬 CANCERFUSION RESEARCH PLATFORM: TIER-1 MULTIMODAL ONCOLOGY ARCHITECTURE
# =====================================================================


class GatedAttentionMIL(nn.Module):
    """
    Gated Attention Multiple Instance Learning (Ilse et al., ICML 2018 / CLAM).
    Learns patch pooling weights; tissue semantics require annotation-based validation.
    """

    def __init__(self, in_features=768, hidden_dim=256, out_dim=64, num_concepts=16):
        super().__init__()
        self.proj = nn.Sequential(
            nn.Linear(in_features, hidden_dim), nn.LayerNorm(hidden_dim), nn.GELU()
        )
        self.attention_V = nn.Sequential(nn.Linear(hidden_dim, 128), nn.Tanh())
        self.attention_U = nn.Sequential(nn.Linear(hidden_dim, 128), nn.Sigmoid())
        self.attention_weights = nn.Linear(128, 1)

        # Unsupervised latent bottleneck (no annotated cellular semantics)
        self.concept_layer = nn.Linear(hidden_dim, num_concepts)
        self.concept_proj = nn.Linear(num_concepts, out_dim)
        self.out_norm = nn.LayerNorm(out_dim)

    def forward(self, x, valid_items=None):
        valid = _item_mask(x, valid_items)
        x = x.masked_fill(~valid.unsqueeze(-1), 0)
        # x: [B, N_patches, 768]
        B, N, D = x.shape
        h = self.proj(x)  # [B, N, 256]

        # Gated Attention Scoring: A = softmax(w^T * (tanh(Vh) * sigm(Uh)))
        A_V = self.attention_V(h)  # [B, N, 128]
        A_U = self.attention_U(h)  # [B, N, 128]
        raw_scores = self.attention_weights(A_V * A_U).squeeze(-1)  # [B, N]

        attn_weights = F.softmax(
            raw_scores.masked_fill(~_safe_attention_mask(valid), float("-inf")), dim=1
        )
        attn_weights = attn_weights.masked_fill(~valid, 0)

        # Attentive Bag Pooling
        bag_repr = torch.bmm(attn_weights.unsqueeze(1), h).squeeze(1)  # [B, 256]

        # Concept Bottleneck Activation
        concepts = torch.sigmoid(self.concept_layer(bag_repr))  # [B, num_concepts]
        out = self.out_norm(self.concept_proj(concepts))  # [B, out_dim]

        out = out.masked_fill(~valid.any(dim=1, keepdim=True), 0)
        return out, concepts, attn_weights


class VolumetricSliceTransformer(nn.Module):
    """
    3D-Aware Positional Slice Transformer for Radiology (MRI DICOM Volumes).
    Uses sequence positions; physical anatomical ordering must be established in preprocessing.
    """

    def __init__(
        self, in_features=512, hidden_dim=256, out_dim=64, max_slices=32, nheads=4
    ):
        super().__init__()
        self.proj = nn.Sequential(
            nn.Linear(in_features, hidden_dim), nn.LayerNorm(hidden_dim), nn.GELU()
        )
        self.pos_embedding = nn.Parameter(torch.randn(1, max_slices, hidden_dim) * 0.02)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=hidden_dim,
            nhead=nheads,
            dim_feedforward=512,
            dropout=0.1,
            activation="gelu",
            batch_first=True,
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=2)
        self.slice_attention = nn.Sequential(
            nn.Linear(hidden_dim, 64), nn.Tanh(), nn.Linear(64, 1)
        )
        self.out_proj = nn.Sequential(
            nn.Linear(hidden_dim, out_dim), nn.LayerNorm(out_dim)
        )

    def forward(self, x, valid_items=None):
        valid = _item_mask(x, valid_items)
        if x.shape[1] > self.pos_embedding.shape[1]:
            raise ValueError("Slice count exceeds positional embedding length.")
        x = x.masked_fill(~valid.unsqueeze(-1), 0)
        # x: [B, S_slices, 512]
        B, S, D = x.shape
        h = self.proj(x)  # [B, S, 256]

        # Add 3D anatomical depth positional embeddings
        h = h + self.pos_embedding[:, :S, :]

        # Cross-slice context encoding
        h_trans = self.transformer(h, src_key_padding_mask=~_safe_attention_mask(valid))

        # Attentive slice pooling
        slice_scores = self.slice_attention(h_trans).squeeze(-1)  # [B, S]
        slice_weights = F.softmax(
            slice_scores.masked_fill(~_safe_attention_mask(valid), float("-inf")), dim=1
        )
        slice_weights = slice_weights.masked_fill(~valid, 0)

        vol_repr = torch.bmm(slice_weights.unsqueeze(1), h_trans).squeeze(1)  # [B, 256]
        out = self.out_proj(vol_repr)  # [B, out_dim]

        out = out.masked_fill(~valid.any(dim=1, keepdim=True), 0)
        return out, slice_weights


class PathwayGenomicSNN(nn.Module):
    """
    Self-Normalizing Neural Network (SNN) with SELU and AlphaDropout for High-Dimensional RNA-Seq.
    Uses an unsupervised latent bottleneck, not annotated biological pathways.
    """

    def __init__(self, in_features=1000, hidden_dim=256, out_dim=64, num_pathways=16):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Linear(in_features, 512),
            nn.SELU(),
            nn.AlphaDropout(p=0.1),
            nn.Linear(512, hidden_dim),
            nn.SELU(),
        )
        # Unsupervised latent layer (no pathway supervision)
        self.pathway_layer = nn.Linear(hidden_dim, num_pathways)
        self.out_proj = nn.Sequential(
            nn.Linear(num_pathways, out_dim), nn.LayerNorm(out_dim)
        )

    def forward(self, x):
        # x: [B, D_rna]
        h = self.encoder(x)
        pathway_activations = torch.sigmoid(self.pathway_layer(h))  # [B, num_pathways]
        out = self.out_proj(pathway_activations)  # [B, out_dim]
        return out, pathway_activations


class ClinicalDenseEncoder(nn.Module):
    """
    LayerNorm Tabular Encoder with Residual Feed-Forward Blocks for Clinical/Demographic Data.
    """

    def __init__(self, in_features=15, out_dim=64):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_features, 128),
            nn.LayerNorm(128),
            nn.GELU(),
            nn.Dropout(0.15),
            nn.Linear(128, out_dim),
            nn.LayerNorm(out_dim),
        )

    def forward(self, x):
        return self.net(x)


class MultiModalCoAttentionTransformer(nn.Module):
    """
    Multi-head fusion with observed availability masks.
    Masking alone does not establish robustness to informative clinical missingness.
    """

    def __init__(self, embed_dim=64, num_modalities=4, nheads=4):
        super().__init__()
        self.embed_dim = embed_dim
        self.num_modalities = num_modalities

        # Learnable Modality Type Embeddings
        self.modality_embeddings = nn.Parameter(
            torch.randn(1, num_modalities, embed_dim) * 0.02
        )

        # Multi-Head Co-Attention
        self.mha = nn.MultiheadAttention(
            embed_dim=embed_dim, num_heads=nheads, dropout=0.1, batch_first=True
        )
        self.norm1 = nn.LayerNorm(embed_dim)

        # Feed-Forward Network
        self.ffn = nn.Sequential(
            nn.Linear(embed_dim, embed_dim * 2),
            nn.GELU(),
            nn.Dropout(0.1),
            nn.Linear(embed_dim * 2, embed_dim),
            nn.LayerNorm(embed_dim),
        )

        # Modality Gating Network (Learns adaptive modality reliability weights)
        self.gating = nn.Sequential(
            nn.Linear(embed_dim, 32), nn.Tanh(), nn.Linear(32, 1)
        )

    def forward(self, modality_tokens, mask):
        # modality_tokens: [B, 4, embed_dim]
        # mask: [B, 4] (1.0 = present, 0.0 = missing)
        B, M, D = modality_tokens.shape

        if mask.shape != (B, M) or M != self.num_modalities:
            raise ValueError("Modality mask and token dimensions must match branches.")
        if not torch.all((mask == 0) | (mask == 1)) or not mask.bool().any(dim=1).all():
            raise ValueError(
                "Each case requires a binary mask and at least one available modality."
            )
        modality_tokens = modality_tokens.masked_fill(~mask.bool().unsqueeze(-1), 0)
        if not torch.isfinite(modality_tokens).all():
            raise ValueError("Available modality tokens must be finite.")

        # Add Modality Type Embeddings
        tokens = modality_tokens + self.modality_embeddings

        # Prepare Key Padding Mask for nn.MultiheadAttention (True = ignored/masked)
        key_padding_mask = mask == 0.0  # [B, 4]

        # Cross-Modal Multi-Head Self-Attention
        attn_out, attn_weights = self.mha(
            tokens, tokens, tokens, key_padding_mask=key_padding_mask
        )
        tokens = self.norm1(tokens + attn_out)
        tokens = self.ffn(tokens) + tokens  # [B, 4, D]

        # Learned Modality Gating (Which modality drives final prognosis)
        gate_logits = self.gating(tokens).squeeze(-1)  # [B, 4]
        # Mask out absent modalities with -inf
        gate_logits = gate_logits.masked_fill(mask == 0.0, float("-inf"))

        modality_importance = F.softmax(gate_logits, dim=1)  # [B, 4]
        # Zero out NaNs if any
        modality_importance = torch.where(
            torch.isnan(modality_importance),
            torch.zeros_like(modality_importance),
            modality_importance,
        )

        # Weighted Fusion
        fused = (tokens * modality_importance.unsqueeze(-1)).sum(dim=1)  # [B, D]

        return fused, modality_importance, attn_weights


class EvidentialSurvivalHead(nn.Module):
    """
    Dempster-Shafer Subjective Logic Head with Dirichlet Distribution.
    Legacy concentration diagnostics; uncertainty interpretation is unvalidated under censoring.
    """

    def __init__(self, in_features=64):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_features, 32), nn.GELU(), nn.Linear(32, 2)
        )

    def forward(self, fused):
        # Evidence non-negative parameterization
        evidence = F.softplus(self.net(fused))
        alpha = evidence + 1.0  # Dirichlet parameters
        S = torch.sum(alpha, dim=1, keepdim=True)  # Dirichlet strength

        # Risk Score (Relative expectation of survival event)
        risk = alpha[:, 1] / S.squeeze(-1)

        # Epistemic Uncertainty: U = K / S (where K=2)
        epistemic_unc = 2.0 / S.squeeze(-1)

        # Aleatoric Uncertainty: Variance of the Dirichlet distribution
        aleatoric_unc = (alpha[:, 0] * alpha[:, 1]) / (
            (S.squeeze(-1) ** 2) * (S.squeeze(-1) + 1.0)
        )

        return risk, epistemic_unc, aleatoric_unc, alpha


class CancerFusionResearchPlatform(nn.Module):
    """
    Research prototype architecture:
    - Gated-MIL Histopathology (WSI)
    - 3D Volumetric Positional Transformer (Radiology MRI)
    - Pathway SNN with Sparse Concept Bottleneck (Genomics RNA)
    - LayerNorm Dense Encoder (Clinical)
    - Multi-Head Cross-Modal Co-Attention Engine
    - Dirichlet Evidential Head (Dual Uncertainty)
    """

    def __init__(self, rna_dim=1000, cli_dim=15, embed_dim=64, max_slices=32):
        super().__init__()
        self.wsi_encoder = GatedAttentionMIL(in_features=768, out_dim=embed_dim)
        self.rad_encoder = VolumetricSliceTransformer(
            in_features=512, out_dim=embed_dim, max_slices=max_slices
        )
        self.rna_encoder = PathwayGenomicSNN(in_features=rna_dim, out_dim=embed_dim)
        self.cli_encoder = ClinicalDenseEncoder(in_features=cli_dim, out_dim=embed_dim)

        self.fusion = MultiModalCoAttentionTransformer(
            embed_dim=embed_dim, num_modalities=4, nheads=4
        )
        self.head = EvidentialSurvivalHead(in_features=embed_dim)

    def forward(self, wsi, rad, rna, cli, mask, wsi_valid=None, rad_valid=None):
        if mask.shape != (wsi.shape[0], 4):
            raise ValueError("Availability mask must have shape [batch, 4].")
        if not torch.all((mask == 0) | (mask == 1)) or not mask.bool().any(dim=1).all():
            raise ValueError(
                "Each case requires a binary mask and at least one available modality."
            )
        inputs = [wsi, rad, rna, cli]
        image_masks = [wsi_valid, rad_valid]
        for index, tensor in enumerate(inputs):
            keep = mask[:, index].bool().reshape((-1,) + (1,) * (tensor.ndim - 1))
            inputs[index] = tensor.masked_fill(~keep, 0)
            if index < 2:
                valid = (
                    _item_mask(inputs[index], image_masks[index])
                    & mask[:, index : index + 1].bool()
                )
                if (mask[:, index].bool() & ~valid.any(dim=1)).any():
                    raise ValueError(
                        "An available image modality must contain at least one valid item."
                    )
                image_masks[index] = valid
                inputs[index] = inputs[index].masked_fill(~valid.unsqueeze(-1), 0)
            if not torch.isfinite(inputs[index]).all():
                raise ValueError("Available feature values must be finite.")
        wsi, rad, rna, cli = inputs
        wsi_valid, rad_valid = image_masks
        # 1. Unimodal Encoding
        f_wsi, wsi_concepts, wsi_patch_attns = self.wsi_encoder(wsi, wsi_valid)
        f_rad, rad_slice_attns = self.rad_encoder(rad, rad_valid)
        f_rna, rna_pathways = self.rna_encoder(rna)
        f_cli = self.cli_encoder(cli)

        # 2. Modality Stack [B, 4, D]
        tokens = torch.stack([f_wsi, f_rad, f_rna, f_cli], dim=1)

        # 3. Safe Cross-Modal Fusion
        fused, modality_weights, cross_attns = self.fusion(tokens, mask)

        # 4. Evidential Prognosis
        risk, epistemic_unc, aleatoric_unc, alpha = self.head(fused)

        explainability = {
            "modality_weights": modality_weights,  # [B, 4]
            "cross_attns": cross_attns,  # [B, 4, 4]
            "wsi_patch_attns": wsi_patch_attns,  # [B, N_patches]
            "wsi_concepts": wsi_concepts,  # [B, 16]
            "rad_slice_attns": rad_slice_attns,  # [B, S_slices]
            "rna_pathways": rna_pathways,  # [B, 16]
            "unimodal_tokens": tokens,  # [B, 4, 64]
            "aleatoric_unc": aleatoric_unc,  # [B]
        }

        return risk, epistemic_unc, modality_weights, explainability, alpha


# =====================================================================
# 📐 COMPOSITE MULTI-TASK LOSS SUITE
# =====================================================================


def continuous_cox_loss(risk, times, events):
    """Batchwise negative Cox partial likelihood with Breslow handling of ties.

    The public interface accepts positive risk scores. Its logarithm is used as
    the Cox log-relative hazard. Censoring is indicated by event=0.
    """
    # Descending times put every risk set in a prefix of this batch.
    order = torch.argsort(times, descending=True)
    ordered_times = times[order]
    ordered_events = events[order].float()
    log_risk = torch.log(risk[order].clamp_min(1e-8))
    log_risk_sum = torch.logcumsumexp(log_risk, dim=0)
    # Breslow ties share the denominator containing all rows with that time.
    _, tie_counts = torch.unique_consecutive(ordered_times, return_counts=True)
    group_end = tie_counts.cumsum(0) - 1
    full_risk_set_logsum = log_risk_sum[group_end.repeat_interleave(tie_counts)]
    return -(
        (log_risk - full_risk_set_logsum) * ordered_events
    ).sum() / ordered_events.sum().clamp_min(1)


def modality_orthogonality_loss(unimodal_tokens, modality_mask=None):
    """
    Mean squared cosine similarity over available off-diagonal modality pairs.
    A single available modality contributes zero; missing branches are excluded.
    """
    B, M, D = unimodal_tokens.shape
    available = (
        torch.ones((B, M), dtype=torch.bool, device=unimodal_tokens.device)
        if modality_mask is None
        else modality_mask.bool()
    )
    if available.shape != (B, M):
        raise ValueError("Orthogonality mask must match token branches.")
    safe_tokens = unimodal_tokens.masked_fill(~available.unsqueeze(-1), 0)
    norm_tokens = F.normalize(safe_tokens, p=2, dim=-1)
    gram = torch.bmm(norm_tokens, norm_tokens.transpose(1, 2))
    pairs = available.unsqueeze(2) & available.unsqueeze(1)
    pairs = pairs & ~torch.eye(M, device=gram.device, dtype=torch.bool).unsqueeze(0)
    return gram.square().masked_fill(~pairs, 0).sum() / pairs.sum().clamp_min(1)


def evidence_strength_penalty(alpha):
    """Mean evidence fraction (S - K) / S for K Dirichlet concentrations.

    This is the pilot's existing evidence penalty. A censored survival uncertainty
    likelihood and clinical validation remain separate research tasks.
    """
    strength = alpha.sum(dim=1)
    return ((alpha - 1.0).sum(dim=1) / (strength + 1e-8)).mean()


def evidential_kl_regularization(alpha, events):
    """Compatibility entry point for the legacy evidence-strength penalty.

    The events argument is retained for existing callers. This function computes
    the same event-independent quantity as the pilot implementation.
    """
    return evidence_strength_penalty(alpha)


def composite_research_loss(
    risk,
    times,
    events,
    alpha,
    unimodal_tokens,
    rna_pathways,
    lambda_edl=0.01,
    lambda_orth=0.05,
    lambda_sparse=0.005,
    modality_mask=None,
):
    """
    Pilot composite objective with an evidence-strength penalty.
    The lambda_edl and l_edl names are retained for existing training callers.
    """
    l_cox = continuous_cox_loss(risk, times, events)
    l_edl = evidence_strength_penalty(alpha)
    l_orth = modality_orthogonality_loss(unimodal_tokens, modality_mask)
    per_case_sparse = torch.sum(torch.abs(rna_pathways), dim=-1)
    if modality_mask is None:
        l_sparse = per_case_sparse.mean()
    else:
        has_rna = modality_mask[:, 2].bool()
        l_sparse = per_case_sparse.masked_fill(
            ~has_rna, 0
        ).sum() / has_rna.sum().clamp_min(1)

    total_loss = (
        l_cox
        + (lambda_edl * l_edl)
        + (lambda_orth * l_orth)
        + (lambda_sparse * l_sparse)
    )
    return total_loss, {
        "l_cox": l_cox.item(),
        "l_edl": l_edl.item(),
        "l_orth": l_orth.item(),
        "l_sparse": l_sparse.item(),
    }
```

```
print("\n" + "="*65)
print("  🚀 IGNITING CANCERFUSION RESEARCH PLATFORM TRAINING ENGINE")
print("="*65)

history = {
    'epoch': [],
    'epoch_seconds': [],
    'total_loss': [],
    'cox_loss': [],
    'edl_loss': [],
    'orth_loss': [],
    'sparse_loss': []
}

from datetime import datetime, timezone
from time import perf_counter
run_dir = Path('/kaggle/working') / datetime.now(timezone.utc).strftime('phase2-%Y%m%d-%H%M%S-%f')
run_dir.mkdir()

model.train()
for epoch in range(1, CFG['epochs']+1):
    epoch_started = perf_counter()
    ep_total_loss = 0
    ep_cox_loss = 0
    ep_edl_loss = 0
    ep_orth_loss = 0
    ep_sparse_loss = 0
    
    successful_updates = 0
    pbar = tqdm(train_dl, desc=f'Epoch {epoch:02d}/{CFG["epochs"]}', leave=False)
    for batch in pbar:
        optimizer.zero_grad(set_to_none=True)
        wsi = batch['wsi'].to(DEVICE)
        rad = batch['rad'].to(DEVICE)
        rna = batch['rna'].to(DEVICE)
        cli = batch['cli'].to(DEVICE)
        mask= batch['mask'].to(DEVICE)
        times = batch['time'].to(DEVICE)
        events= batch['event'].to(DEVICE)
        
        risk, unc, weights, xai, alpha = model(wsi, rad, rna, cli, mask, batch['wsi_valid'].to(DEVICE), batch['rad_valid'].to(DEVICE))
        loss, loss_dict = composite_research_loss(
            risk, times, events, alpha, xai['unimodal_tokens'], xai['rna_pathways'], modality_mask=mask
        )
        
        if not torch.isfinite(loss).item():
            raise ValueError(f"Non-finite training loss during epoch {epoch}. Review the inputs and training configuration.")
        loss.backward()
        optimizer.step()
        successful_updates += 1
        ep_total_loss += loss.item()
        ep_cox_loss += loss_dict['l_cox']
        ep_edl_loss += loss_dict['l_edl']
        ep_orth_loss += loss_dict['l_orth']
        ep_sparse_loss += loss_dict['l_sparse']

    if successful_updates == 0:
        raise ValueError("The training loader produced no batches.")
    num_batches = successful_updates
    if DEVICE.type == 'cuda':
        torch.cuda.synchronize(DEVICE)
    elapsed = perf_counter() - epoch_started
    history['epoch'].append(epoch)
    history['epoch_seconds'].append(elapsed)
    history['total_loss'].append(ep_total_loss / num_batches)
    history['cox_loss'].append(ep_cox_loss / num_batches)
    history['edl_loss'].append(ep_edl_loss / num_batches)
    history['orth_loss'].append(ep_orth_loss / num_batches)
    history['sparse_loss'].append(ep_sparse_loss / num_batches)
    
    print(f"Epoch {epoch:02d}/{CFG['epochs']:02d} | Total Composite Loss: {ep_total_loss/num_batches:.4f} | Cox Hazard Loss: {ep_cox_loss/num_batches:.4f} | {elapsed:.1f} s")

torch.save(model.state_dict(), run_dir / 'CancerFusion_Phase2_Weights.pth')
pd.DataFrame(history).to_csv(run_dir / 'training_history.csv', index=False)
print(f"Training checkpoint and history saved to {run_dir}")
```

```
"""Evaluate the trained checkpoint once on observed, right-censored outcomes."""

import hashlib
from lifelines import CoxPHFitter
from lifelines.exceptions import ConvergenceError
from lifelines.utils import concordance_index

output_dir = run_dir
figure_dir = output_dir / "paper_figures"
figure_dir.mkdir(parents=True, exist_ok=True)
modalities = ("wsi", "rad", "rna", "cli")


def observed_c_index(rows, risk_column):
    if rows.empty or not rows["event"].any():
        return None
    try:
        return float(concordance_index(rows["time"], -rows[risk_column], rows["event"]))
    except ZeroDivisionError:
        return None


def concordance_summary(rows, risk_columns, reference_column=None):
    """Use identical patient bootstrap draws for every score in a comparison."""
    rows = rows.reset_index(drop=True)
    points = {key: observed_c_index(rows, key) for key in risk_columns}
    samples = {key: [] for key in risk_columns}
    differences = {key: [] for key in risk_columns if reference_column and key != reference_column}
    rng = np.random.default_rng(CFG["seed"])
    if len(rows) and any(point is not None for point in points.values()):
        for _ in range(CFG["bootstrap_replicates"]):
            draw = rows.iloc[rng.integers(len(rows), size=len(rows))]
            scores = {key: observed_c_index(draw, key) for key in risk_columns}
            for key, score in scores.items():
                if score is not None:
                    samples[key].append(score)
                    if key in differences and scores[reference_column] is not None:
                        differences[key].append(scores[reference_column] - score)

    def interval(values):
        # Resamples without comparable pairs are visible in the reported count.
        if len(values) < 2:
            return None
        return np.percentile(values, [2.5, 97.5]).tolist()

    summary = {
        "patients": len(rows),
        "observed_events": int(rows["event"].sum()),
        "bootstrap_replicates_requested": CFG["bootstrap_replicates"],
        "scores": {},
    }
    for key, point in points.items():
        stats = {
            "harrell_c": point,
            "bootstrap_95_ci": interval(samples[key]),
            "bootstrap_replicates_valid": len(samples[key]),
        }
        if key in differences:
            reference = points[reference_column]
            stats.update(
                reference=reference_column,
                reference_minus_model_c=reference - point if point is not None and reference is not None else None,
                paired_difference_bootstrap_95_ci=interval(differences[key]),
                paired_difference_replicates_valid=len(differences[key]),
            )
        summary["scores"][key] = stats
    return summary


@torch.inference_mode()
def evaluate_validation(loader):
    model.eval()
    records = []
    for batch in tqdm(loader, desc="Validation inference"):
        tensors = [batch[key].to(DEVICE) for key in modalities]
        available = batch["mask"].to(DEVICE)
        if not (available > 0).any(dim=1).all().item():
            raise ValueError("A validation case has no model inputs. Review the modality audit.")
        risk, uncertainty, weights, _, _ = model(*tensors, available)
        if not all(torch.isfinite(value).all().item() for value in (risk, uncertainty, weights)):
            raise ValueError("Model produced non-finite validation outputs.")
        complete = (available > 0).all(dim=1)
        digest = {}
        if complete.any().item():
            complete_inputs = [value[complete] for value in tensors]
            for column, modality in enumerate(modalities):
                mask = available[complete].clone()
                mask[:, column] = 0
                masked_risk = model(*complete_inputs, mask)[0]
                if not torch.isfinite(masked_risk).all().item():
                    raise ValueError(f"Model produced non-finite scores with {modality} withheld.")
                result = torch.full_like(risk, float("nan"))
                result[complete] = masked_risk
                digest[f"without_{modality}_risk"] = result.cpu().numpy()
        frame = pd.DataFrame({
            "patient_id": batch["patient_id"],
            "time": batch["time"].cpu().numpy(),
            "event": batch["event"].cpu().numpy().astype(int),
            "fusion_risk": risk.cpu().numpy(),
            "evidence_uncertainty": uncertainty.cpu().numpy(),
        })
        for column, key in enumerate(modalities):
            frame[f"has_{key}"] = available[:, column].cpu().numpy().astype(bool)
            frame[f"weight_{key}"] = weights[:, column].cpu().numpy()
        for key, values in digest.items():
            frame[key] = values
        records.append(frame)
    predictions = pd.concat(records, ignore_index=True)
    for key in modalities:
        name = f"without_{key}_risk"
        if name not in predictions:
            predictions[name] = np.nan
    if predictions["patient_id"].duplicated().any():
        raise ValueError("The validation export contains duplicate patient identifiers.")
    return predictions


predictions = evaluate_validation(val_dl)
paired = predictions[predictions[[f"has_{key}" for key in modalities]].all(axis=1)].copy()
metrics = {
    "evaluation_scope": "Internal held-out validation, observed time-to-event and right censoring",
    "checkpoint_sha256": hashlib.sha256((output_dir / "CancerFusion_Phase2_Weights.pth").read_bytes()).hexdigest(),
    "seed": CFG["seed"],
    "training_patients": len(train_df),
    "training_observed_events": int(train_df["event"].sum()),
    "modality_counts": {key: int(predictions[f"has_{key}"].sum()) for key in modalities},
    "all_validation_cases": concordance_summary(predictions, ["fusion_risk"]),
    "paired_four_modality_cases": concordance_summary(
        paired, ["fusion_risk", *(f"without_{key}_risk" for key in modalities)], "fusion_risk"
    ),
}

# A baseline can only use explicitly reviewed predictors known at the prediction date.
# Features are selected and standardized using training rows alone.
approved = CFG["clinical_baseline_columns"]
predictions["clinical_cox_risk"] = np.nan
if not approved:
    metrics["clinical_cox_baseline"] = {"status": "not_run", "reason": "Configure reviewed clinical_baseline_columns."}
else:
    unknown = set(approved) - set(cli_df.columns)
    if unknown or len(approved) != len(set(approved)):
        raise ValueError(f"Clinical baseline columns are unknown or duplicated: {sorted(unknown)}")
    training = train_df.set_index("Patient_ID").join(cli_df[approved], how="inner")
    test = predictions[predictions["has_cli"]].set_index("patient_id").join(cli_df[approved], how="inner")
    varying = training[approved].std(ddof=0)
    used = varying[varying.gt(0)].index.tolist()
    if test.empty or not used or not training["event"].any():
        metrics["clinical_cox_baseline"] = {"status": "not_run", "reason": "Insufficient eligible clinical inputs or training events."}
    else:
        mean, scale = training[used].mean(), varying[used]
        clinical = CoxPHFitter(penalizer=0.1)
        try:
            clinical.fit(
                ((training[used] - mean) / scale).join(training[["time", "event"]]),
                duration_col="time", event_col="event",
            )
        except ConvergenceError as exc:
            metrics["clinical_cox_baseline"] = {"status": "failed_to_converge", "reason": str(exc)}
        else:
            risk = clinical.predict_log_partial_hazard((test[used] - mean) / scale)
            predictions["clinical_cox_risk"] = predictions["patient_id"].map(risk)
            matched = predictions.dropna(subset=["clinical_cox_risk"])
            metrics["clinical_cox_baseline"] = {
                "status": "evaluated",
                "training_patients": len(training),
                "training_events": int(training["event"].sum()),
                "columns": used,
                **concordance_summary(matched, ["fusion_risk", "clinical_cox_risk"], "fusion_risk"),
            }

predictions.to_csv(output_dir / "validation_predictions.csv", index=False, na_rep="")
(output_dir / "complete_metrics_summary.json").write_text(
    json.dumps(metrics, indent=2, allow_nan=False), encoding="utf-8"
)

fig, ax = plt.subplots(figsize=(8, 5))
for key, label in (("total_loss", "Total"), ("cox_loss", "Batch Cox"),
                   ("edl_loss", "Evidence penalty"), ("orth_loss", "Orthogonality")):
    ax.plot(history["epoch"], history[key], label=label)
ax.set(title="Training losses", xlabel="Epoch", ylabel="Loss")
ax.legend()
fig.tight_layout()
fig.savefig(figure_dir / "Figure1_Loss_Convergence.png", dpi=180)
plt.close(fig)

fig, ax = plt.subplots(figsize=(8, 5))
weights = predictions[[f"weight_{key}" for key in modalities]]
ax.bar([key.upper() for key in modalities], weights.mean(), yerr=weights.std(ddof=0), capsize=4)
ax.set(title="Validation pooling weights (mean and SD)", ylabel="Pooling weight", ylim=(0, 1))
fig.tight_layout()
fig.savefig(figure_dir / "Figure6_Modality_Importance.png", dpi=180)
plt.close(fig)

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(predictions["fusion_risk"], predictions["evidence_uncertainty"])
ax.set(title="Observed validation outputs", xlabel="Raw relative risk score", ylabel="Evidence uncertainty score")
fig.tight_layout()
fig.savefig(figure_dir / "Figure8_Risk_and_Evidence_Uncertainty.png", dpi=180)
plt.close(fig)

scores = metrics["paired_four_modality_cases"]["scores"]
names = [key for key, result in scores.items() if result["harrell_c"] is not None]
if names:
    fig, ax = plt.subplots(figsize=(9, 5))
    ax.barh(names[::-1], [scores[key]["harrell_c"] for key in names[::-1]])
    ax.set(xlim=(0, 1), xlabel="Harrell C-index", title="Comparison on the same four-modality cases")
    fig.tight_layout()
    fig.savefig(figure_dir / "Figure11_Paired_Ablations.png", dpi=180)
    plt.close(fig)

print(json.dumps(metrics, indent=2, allow_nan=False))
if metrics["all_validation_cases"]["scores"]["fusion_risk"]["harrell_c"] is None:
    print("Validation contains no comparable observed-event pairs. Clinical discrimination is unavailable.")
print("Saved validation predictions, survival metrics and computed figures.")
```
